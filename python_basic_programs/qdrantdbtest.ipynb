```python

import os
import base64
from pathlib import Path
from dotenv import load_dotenv
from openai import OpenAI
from qdrant_client import QdrantClient, models

load_dotenv()

EURI_API_KEY = os.getenv("EURI_API_KEY")
QDRANT_URL = os.getenv("QDRANT_URL")
QDRANT_API_KEY = os.getenv("QDRANT_API_KEY")

euri = OpenAI(
    api_key=EURI_API_KEY,
    base_url="https://api.euron.one/api/v1/euri"
)

qdrant = QdrantClient(
    url=QDRANT_URL,
    api_key=QDRANT_API_KEY
)

COLLECTION = "sudh_image_collection"
MODEL = "gemini-embedding-2-preview"

IMAGE_FOLDER = "images"

SUPPORTED_FORMATS = {".jpg", ".jpeg", ".png", ".webp"}

def embed_image(image_path):

    with open(image_path, "rb") as f:
        image_data = f.read()

    encoded_image = base64.b64encode(image_data).decode("utf-8")

    response = euri.embeddings.create(
        model=MODEL,
        input=encoded_image
    )

    return response.data[0].embedding


def process_multiple_images(folder_path):

    results = []

    image_files = [
        p for p in Path(folder_path).iterdir()
        if p.is_file() and p.suffix.lower() in SUPPORTED_FORMATS
    ]

    for index, image_path in enumerate(image_files, start=1):

        try:
            embedding = embed_image(image_path)

            results.append({
                "id": index,
                "image_name": image_path.name,
                "embedding": embedding
            })

            print(f"Processed: {image_path.name}")

        except Exception as e:
            print(f"Error processing {image_path.name}: {e}")

    return results


# Process all images
image_results = process_multiple_images(IMAGE_FOLDER)

print("Total images processed:", len(image_results))
```
